### **Handle Missing Values**

Replace the missing value in the 2025-2026 DataFrame's <code>local_authority</code> column with "Unknown". According to the Arts Council documentation, the recipient may operate across multiple local authorities, so assigning a specific authority could misrepresent the data.

```python
# Replace missing value in the 'local_authority' column with 'Unknown'
df_26['local_authority'] = df_26['local_authority'].fillna('Unknown')

# Check for remaining missing values in df_26
df_26.isna().sum()
```

Replace missing values in the <code>time_limited_priority</code> column of both the 2023-2024 and 2025-2026 DataFrames with "Not Time Limited Priority". Arts Council documentation confirms that time-limited priorities do not apply to all awards. This replacement preserves all records while distinguishing awards without a time-limited priority from those with a specified priority.

```python
# Replace missing values in the 'time_limited_priority' columns
for df in [df_24, df_26]:
   df['time_limited_priority'] = (
       df['time_limited_priority']
       .fillna('Not Time-Limited Priority'))

# Check for remaining missing values in df_24 and df_26
print("Missing values in df_24 after replacement:")
print(df_24.isna().sum())
print("\nMissing values in df_26 after replacement:")
print(df_26.isna().sum())
```

### **Award Amount**

Similar to standardising column names in the previous section, use <code>.replace()</code> and <code>.strip()</code> to remove unwanted symbols and leading or trailing whitespace from numerical values. This helps prevent <code>ValueError</code> exceptions caused by invalid string literals during conversion. Convert the cleaned string values to the integer data type using <code>.astype(int)</code>.

```python
# Clean and convert the 'award_amount' column in df_24
df_24["award_amount"] = (df_24["award_amount"]
                         .str.replace(',','')
                         .str.replace('£','')
                         .str.strip()
                         .astype(int))


# Clean and convert the 'award_amount' column in df_26
df_26["award_amount"] = (df_26["award_amount"]
                         .str.replace(',','')
                         .str.replace('£','')
                         .str.strip()
                         .astype(int))


# Check the data type of the 'award_amount' column
print("Data type df_24:", df_24["award_amount"].dtype) 
print("Data type df_26:", df_26["award_amount"].dtype) 
```

### **Date-Time**

Convert the <code>award_date</code> column from the string data type to the datetime data type in both DataFrames using <code>pd.to_datetime()</code>. Check that the conversion is successful to ensure dates can be processed correctly.

Create a separate year column by extracting the year from <code>award_date</code> using <code>.dt.year</code>. This enables easier filtering, grouping and year-on-year comparisons in the EDA and dashboard visualisations.

```python
# Convert the award_date column to datetime for both DataFrames
for data in (df_24, df_26):
    data['award_date'] = pd.to_datetime(
        data['award_date'],
        errors='coerce'
    )
    # Extract the award year from the datetime values
    data['award_year'] = (
        data['award_date']
        .dt.year
        .astype('Int64')
    )
```

Standardise the date format and handle invalid dates safely by using <code>errors='coerce'</code> within <code>pd.to_datetime()</code>. Invalid dates are converted to <code>NaT</code> (Not a Time), allowing the conversion to continue without raising an error. These missing date-time values can then be identified and investigated during data cleaning.

```python
# Check the first few rows of df_24 to verify the changes
df_24.head()   
```

Initially, the built-in Copilot modified the code to remove the timestamp using <code>.normalize()</code>. However, this method only resets the time to midnight and does not remove the time component. The appropriate method is <code>.dt.date</code>, which extracts the date without the timestamp.

As extracting the year requires the datetime data type, create the year column using <code>.dt.year</code> before applying <code>.dt.date</code>. This preserves the ability to extract the year while removing the timestamp from the date column.

```python
# Remove time from award_date column in df_24 and df_26
df_24['award_date'] = df_24['award_date'].dt.date
df_26['award_date'] = df_26['award_date'].dt.date
```

### **Categorical**

Categorical columns contain a finite set of unique values representing distinct groups or categories. Use <code>.astype("category")</code> to convert string columns into the categorical data type, improving data consistency and supporting efficient grouping and analysis in subsequent notebooks.

```python
# Group categorical columns for conversion to 'category' data type
categorical_cols = [
    "main_discipline",
    "strand",
    "time_limited_priority",
    "ace_area"
]

# Convert the categorical columns
for df in [df_24, df_26]:
    for col in categorical_cols:
        if col in df.columns:
            df[col] = df[col].astype("category")
```
