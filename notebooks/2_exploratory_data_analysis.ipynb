## <u>**Outlier Inspection**</u>

Winsorization is a statistical technique used to reduce the influence of extreme values. Instead of removing observations, values below the 5th percentile are capped at the 5th percentile, and values above the 95th percentile are capped at the 95th percentile. This preserves the number of observations but replaces extreme values with the respective percentile thresholds.

The initial plots are heavily influenced by a small number of very large awards, making the central tendency and overall distribution difficult to interpret. Apply winsorization to visualise the typical range of award amounts and assess how extreme values affect the distribution.

```python
# Import the FuncFormatter for custom tick formatting
from matplotlib.ticker import FuncFormatter

# Calculate the 5th and 95th percentile thresholds for winsorisation
lower_limit = float(amounts.quantile(0.05))
upper_limit = float(amounts.quantile(0.95))

# Count how many awards fall above the 95th percentile threshold
excluded = int((amounts > upper_limit).sum())

# Calculate the median award value for comparison
median_amount = float(amounts.median())

# Apply notebook styling for consistency across plots
sns.set_theme(style="whitegrid", context="notebook")

# Create a figure and axis for the histogram
fig, ax = plt.subplots(figsize=(11, 6))

# Plot a histogram of award amounts capped at the 95th percentile
sns.histplot(
    x=amounts[amounts <= upper_limit],  # Only include values up to the 95th percentile
    bins=40,                           # Number of bins for the distribution
    kde=True,                          # Overlay a kernel density estimate
    color="#3274A1",                   # Blue histogram bars
    alpha=0.75,                        # Slight transparency
    edgecolor="white",                 # White bar borders
    ax=ax,                             # Draw on the current axis
)

# Add a vertical line at the 5th percentile
ax.axvline(
    lower_limit,                       # Use float for type safety
    color="#009E73",                   # Green line
    linestyle=":",                     # Dotted line
    linewidth=2,
    label=f"5th percentile: £{lower_limit:,.0f}",
)

# Add a vertical line at the median
ax.axvline(
    median_amount,                     # Use float for type safety
    color="#D55E00",                   # Orange line
    linestyle="--",                    # Dashed line
    linewidth=2,
    label=f"Median: £{median_amount:,.0f}",
)

# Set plot title and axis labels
ax.set_title("Distribution of Award Amounts", fontsize=16, weight="bold", pad=15)
ax.set_xlabel("Award amount (£)")
ax.set_ylabel("Number of awards")

# Format x-axis tick labels as currency
ax.xaxis.set_major_formatter(FuncFormatter(lambda value, _: f"£{value:,.0f}"))

# Add an annotation explaining the exclusion of extreme values
ax.text(
    0.98,
    0.95,
    f"Displaying awards up to the 95th percentile\n"
    f"{excluded} extreme value(s) excluded from view",
    transform=ax.transAxes,
    ha="right",
    va="top",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", alpha=0.85),
)

# Add legend and grid styling for readability
ax.legend(frameon=False)
ax.grid(axis="y", alpha=0.3)
ax.grid(axis="x", visible=False)

# Remove top/right spines for a cleaner plot
sns.despine()

# Adjust layout to avoid clipping
plt.tight_layout()

# Display the final chart
plt.show()
```

The histogram shows that the award amount distribution is positively skewed, with most awards concentrated in the lower-to-middle range and a long right-hand tail containing substantially larger awards.

## <u>**Geographic Hypothesis Test**</u>

Descriptive statistics alone cannot establish whether differences in mean award amounts across ACE regions are statistically significant. Welch’s ANOVA is therefore used to test for differences in regional mean award amounts without assuming equal population variances.

### **Hypotheses**

**Null Hypothesis** - Mean award amoutns are equal across all five ACE regions.

**Alternative Hypothesis** - At least one ACE region has a different mean award amount. 

The significance level is set at (\alpha = 0.05). If the p-value is below 0.05, the null hypothesis is rejected, indicating statistically significant evidence that mean award amounts differ across at least one region.

Welch’s ANOVA identifies whether at least one regional mean differs but does not identify which regions differ. Post-hoc tests would be required to determine specific pairwise differences.

</b>

### **Welch's ANOVA**

Welch’s ANOVA was selected to compare mean award amounts across the five ACE regions because it does not require each region to have the same variance. This is important because the spread of award values may differ between regions.

```python

# Import the ANOVA function for Welch's ANOVA test
from statsmodels.stats.oneway import anova_oneway

# Create a focused DataFrame with only the ACE area and award amount columns
analysis_df = df_combined[["ace_area", "award_amount"]].copy()

# Convert award amounts to numeric values, treating invalid entries as missing
analysis_df["award_amount"] = pd.to_numeric(
    analysis_df["award_amount"], errors="coerce"
)

# Drop rows where either the ACE area or award amount is missing
analysis_df = analysis_df.dropna(
    subset=["ace_area", "award_amount"]
)

# Exclude zero or negative award values since the hypothesis compares mean award amounts
analysis_df = analysis_df[
    analysis_df["award_amount"] > 0
]

# Summarise award amount by ACE region to inspect group sizes and variation
display(
    analysis_df.groupby("ace_area")["award_amount"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

# Split the DataFrame into a list of arrays one for each ACE region
groups = [
    group["award_amount"].to_numpy()
    for _, group in analysis_df.groupby("ace_area")
]

# Validate that all five ACE regions are present and each has at least two observations
if len(groups) != 5 or any(len(g) < 2 for g in groups):
    raise ValueError(
        "Check that all five ACE regions have at least "
        "two valid award records."
    )

# Run Welch's ANOVA, which does not assume equal variances across groups
result = anova_oneway(groups, use_var="unequal")

# Print the test statistic and p-value for reporting
print(f"Welch's ANOVA statistic: {result.statistic:.4f}")
print(f"P-value: {result.pvalue:.6g}")

# Set the significance threshold for hypothesis testing
alpha = 0.05

# Decision rule reject the null hypothesis if the p-value is below alpha
if result.pvalue < alpha:
    print(
        "Reject H0: evidence that mean award amounts "
        "differ across at least one ACE region."
    )
else:
    print(
        "Fail to reject H0: insufficient evidence that "
        "mean award amounts differ across ACE regions."
    )
```

Welch’s ANOVA returned a test statistic of 3.4732 and a p-value of 0.0077582. As the p-value is below the 0.05 significance threshold, the null hypothesis was rejected. This suggests that mean award amounts are not equal across all five ACE regions, although the test does not identify which regions differ.

These findings indicate regional variation in recorded award amounts but do not establish its causes or demonstrate funding bias. As the dataset includes only successful awards, it cannot assess regional differences in applicants’ chances of receiving funding. Further analysis would be required to determine whether the observed variation reflects inequity in the funding process.
